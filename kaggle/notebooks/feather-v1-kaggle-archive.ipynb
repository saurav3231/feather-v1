```python
import struct
gguf = 'models/kaggle/feather-v1-kaggle.gguf'
raw = open(gguf, 'rb').read(32)
print('magic:', raw[:4], '| version:', struct.unpack('<I', raw[4:8])[0])
print('tensor_count/kv:', struct.unpack('<QQ', raw[8:24]))
```

```python
# tokenizer.json + config.json + tar members
import json, tarfile
tok = json.load(open('models/kaggle/tokenizer.json', encoding='utf-8'))
print('tokenizer vocab entries:', len(tok['model']['vocab']))
cfg = json.load(open('models/kaggle/config.json', encoding='utf-8'))
print('config dim/vocab:', cfg.get('dim'), cfg.get('vocab_size'))
with tarfile.open('models/kaggle/feather-v1-kaggle.tar.gz') as tf:
    for m in tf.getnames()[:8]:
        print('  ', m)
```

```python
# npz round-trip
import numpy as np
z = np.load('models/kaggle/feather-v1-kaggle.pt', allow_pickle=False)
print('members:', list(z))
print('logit_projection:', z['logit_projection'].shape)
print('npz header config dim:', json.loads(bytes(z['config']).decode('utf-8'))['dim'])
```
